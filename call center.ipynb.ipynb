```python
import pandas as pd
```

```python
df = pd.read_csv("call_center.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df = df.drop(['Unnamed: 9','Unnamed: 10'], axis = 1, inplace = False)
```

```python
df.duplicated().sum()
```

```python
df.columns
```

```python
df.dtypes
```

```python
df.describe()
```

```python
df["Speed of Answer"] = df["Speed of Answer"].fillna(0)
```

```python
df["AvgTalkDuration"] = df["AvgTalkDuration"].fillna("00:00:00")
```

```python
df["Satisfaction rating"] = df["Satisfaction rating"].fillna(0)
```

```python
#Python Analysis (EDA)
#total number of calls
df.shape
```

```python
#Total calls
len(df)
```

```python
#Total Agents
df["Agent"].unique()
```

```python
#Call handled by each agent
df["Agent"].value_counts()  
#Business insights : Identifies which agent handled the most calls
```

```python
#Call by department
df["Department"].value_counts()
#Bussiness Insights: shows which department receives the highest cals
```

```python
#Answered vs non answered
df["Answered (Y/N)"].value_counts()
#Bussiness Insights : percentage of answered and missed calls
```

```python
#Resolved vs Unresolved calls
df["Resolved"].value_counts()
# Bussiness Insights : shows how effective the call center is at resolving customer issue
```

```python
#Average speed of answer
df["Speed of Answer"].mean()
#Bussiness Insights : lower average speed means customer wait less 
```

```python
df["AvgTalkDuration"].head(10)
```

```python
#converting it to time format before calculating average
df["AvgTalkDuration"] = pd.to_timedelta(df["AvgTalkDuration"])
```

```python
#Average talk duration 
df["AvgTalkDuration"].mean()
#Bussiness Insights : helps understand how long agent speed with customers
```

```python
#Average satisfication rating
df["Satisfaction rating"].mean()
#Bussiness Insights : measures overall customer satisfication
```

```python
#Highest satisfaction rating
df["Satisfaction rating"].max()
```

```python
#Lowest satisfaction rating
df["Satisfaction rating"].min()
```
